```python
print("\nChoose Career Goal")

print("1. AI Engineer")
print("2. Data Scientist")
print("3. Python Developer")

career = input("Enter choice: ")
```

```python
skills = input(
    "\nEnter your current skills separated by commas: "
)
```

```python
if career == "1":

    required_skills = [
        "Python",
        "NumPy",
        "Pandas",
        "Machine Learning",
        "Deep Learning",
        "Git"
    ]

elif career == "2":

    required_skills = [
        "Python",
        "Statistics",
        "Pandas",
        "Data Visualization",
        "Machine Learning"
    ]

else:

    required_skills = [
        "Python",
        "Functions",
        "OOP",
        "Git",
        "APIs"
    ]
```

```python
user_skills = skills.split(",")

user_skills = [
    skill.strip()
    for skill in user_skills
]
```

```python
print("\n✅ Skills You Already Have\n")

for skill in user_skills:
    print(skill)
```

```python
missing_skills = []

for skill in required_skills:

    if skill not in user_skills:

        missing_skills.append(skill)
```

```python
print("\n📚 Skills To Learn\n")

for skill in missing_skills:

    print("❌", skill)
```

```python
if len(missing_skills) > 0:

    print(
        "\n🎯 Recommended Next Skill:"
    )

    print(
        missing_skills[0]
    )
```

```python
print("\n🚀 Suggested Projects")

if "Machine Learning" in missing_skills:

    print("- Student Score Predictor")

if "Deep Learning" in missing_skills:

    print("- Hand Gesture Recognition")

if "Git" in missing_skills:

    print("- GitHub Portfolio Setup")
```
