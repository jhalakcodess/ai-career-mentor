```python
print("\nChoose Career Goal")

print("1. AI Engineer")
print("2. Data Scientist")
print("3. Python Developer")

career = input("Enter choice: ")
```

```python
skills = input(
    "\nEnter your current skills separated by commas: "
)
```

```python
if career == "1":

    required_skills = [
        "Python",
        "NumPy",
        "Pandas",
        "Machine Learning",
        "Git",
        "Deep Learning"
    ]

elif career == "2":

    required_skills = [
        "Python",
        "Statistics",
        "Pandas",
        "Data Visualization",
        "Machine Learning"
    ]

else:

    required_skills = [
        "Python",
        "Functions",
        "OOP",
        "Git",
        "APIs"
    ]
```

```python
user_skills = skills.split(",")

user_skills = [
    skill.strip()
    for skill in user_skills
]
```

```python
missing_skills = []

for skill in required_skills:

    if skill not in user_skills:

        missing_skills.append(skill)
```

```python
print("\n📚 PERSONALIZED ROADMAP\n")

step = 1

for skill in missing_skills:

    print(
        f"Step {step} → Learn {skill}"
    )

    step += 1
```

```python
print("\n🚀 PROJECTS TO BUILD\n")

if career == "1":

    print("1. Student Score Predictor")
    print("2. AI Study Planner")
    print("3. AI Interview Trainer")
    print("4. AI Career Mentor")

elif career == "2":

    print("1. Data Analysis Dashboard")
    print("2. Sales Prediction Model")
    print("3. Recommendation System")

else:

    print("1. Calculator App")
    print("2. Expense Tracker")
    print("3. Weather App")
```

```python
months = len(missing_skills)

print(
    f"\n⏳ Estimated Duration: {months}-{months+2} Months"
)
```
